## Validating the weather forecast days input

```python
# Creating functon for the weather forecast days input

def getting_weather_forecast_days ():
    
    while True:
        weather_value = input(
            "Please enter the number of days from 1 to 7 [the default will automatically be 3]:"
        ).strip()


        if weather_value == "":
            return 3 

        try:
            weather_days = int(weather_value)

            if 1 <= weather_days <= 7:
                return weather_days



        except ValueError:
            pass

        print("Please enter a whole number from 1 to 7.")
```

- The following functions for `getting_weather_forecast_days` is asking the user to enter some information as required.

## Getting the city coordinates

```python
# Creating functon for city coordinates

def geocode_city_coordinates(find_city):

    geocode_parameters = {
        "name": find_city,
        "count": 1,
        "language": "en",
        "format": "json"
    }


    getting_geocode_response = requests.get(
        geocoding_url,
        params=geocode_parameters,
        timeout=20
    )


    getting_geocode_response.raise_for_status()

    getting_data = getting_geocode_response.json()

    getting_results = getting_data.get("results", [])

    if not getting_results:
        raise ValueError(
            f"There is no location found for '{find_city}'."
        )

    return getting_results[0]
```

Based on the function for `gecode_city_coordinates` this will able to retrive the information that the user requests the city they have inputted.

The API will be able to do the following:

- Based on the user has inputted based on their city of choice and the API should be able to gather the search and give information.
- Having the ability for the API to have have one results as required.
- The language is set as English for comphrension and better outlook.
- Setting the result to be able return as JSON for simplisitc outlook.
- The `request.get()` gives the ability to conncect to the API
- The `timeout = 20` explains the request to be able to stop running after 20 seconds if the API does not respond to the request.
- `getting_geocode_response.raise_for_status()` overviews if there any HTTP errors present and usally the response is transformed into Python data by using `getting_data = getting_geocode_response.json()`.
- The following city that the user has inputted and requested using the `getting_results = getting_data.get("results", [])` if the following city that was request has not been found then it will output has no location found for select city.
- The return value for the user that has selected for input city will return it's value as follows.

## Dowloading the weather information as follows:

```python
# Creating a function for getting the weather information 

def fetching_weather_info(longitude, latitude, forecast_days):
    
    weather_parameters = {
        'latitude': latitude,
        'longitude': longitude,

       "current": (
        "temperature_2m,"
        "relative_humidity_2m,"
        "apparent_temperature,"
        "precipitation,"
        "weather_code,"
        "wind_speed_10m"       
    ),


     "daily": (
        "weather_code,"
        "temperature_2m_max,"
        "temperature_2m_min,"
        "precipitation_probability_max,"
        "wind_speed_10m_max",
        "sunrise,"
        "sunset"
    ), 


    "timezone": "auto",
    "forecast_days": forecast_days
}

    getting_forecast_info = requests.get(
        forecast_url,
        params=weather_parameters,
        timeout=20
    )

    getting_forecast_info.raise_for_status()

    return getting_forecast_info.json()
```

- The following function for `fetching_weather_info` gets the following information such as the longitude, latitude and the number of days within the  weather forecast.
- The parameters explains what Open Metro to retrive the following weather information when needed.
- The following requests within the parameters calls such as the current weather and the daily weather.
- The following `timwzone` usally calls for the location based on the users current local time.
- Te following `getting_forecast_info` is send the API for a request then the return JSON data will have the ability send it towards the progam. 

## Cleaning the filename

```python
# This will remove any characters that can cause errors towards the filename

def cleaning_filename_section(filename_text):
    cleaned_filename_text = re.sub(
        r"[^A-Za-z0-9_-]+",
        "_",
        filename_text.strip()
    )

    return cleaned_filename_text.strip("_") or "getting_location"
```

- This function `cleaning_filename_section` make sure that the city that the user has inputed when required and it is safe to use as a filename when outputting and converting it to a word document.
- The following regular expression subsititutes any unsual characters anf underscores that are not suitable for filename.
- Therefore the function is returned with a cleaned text that is applicable for filename.

## Generating a date filename 

```python
# Creating a function for filename report with a date 

def generate_report_path(find_city):
    current_date_section = datetime.now().strftime("%Y-%m-%d") # Year / Month / Day Format

    filename_generate = (
        f"Weather_report_info_"
        f"{cleaning_filename_section(find_city)}_"
        f"{current_date_section}"
    )


    report_section_info_path = os.path.join(
        OUTPUT_DIR, f"{filename_generate}.docx"
    )

    counter = 2
        
    while (
        os.path.exists(report_section_info_path)
        or os.path.exists(os.path.join(LOGS_DIR, os.path.basename(report_section_info_path)))
    ):
        report_section_info_path = os.path.join(
            OUTPUT_DIR,
            f"{filename_generate}_{counter}.docx"
        )

        counter += 1

    return report_section_info_path
```

- The function such as `generate_report_path` give the ability to create a report filename as for the current date section is been created.
- The following layout of the date is as follows year/month/day then the filename of the selected city is inputted by the user, the filename is being created as well as the document extenstion format is added towards the filename.
- The loop method reviews on either the following file that is presented that is already exists.
- If the following filename exists in the data, therefore the filename will add a number as this will prevent from overwriting the file.

## Creating the word document and adding the report heading for the Word document for the Weather report

```python
# Creating a function where you can create a Word document for the weather report and creating a font

def creating_word_doc_report(getting_user_name, getting_location, weather_info_section): # weather # getting_location
    
    find_city = getting_location["name"]
    get_country = getting_location.get("country", "")

    report_section_info_path = generate_report_path(find_city)

    get_document = Document()

    get_document.styles["Normal"].font.name = "Calibri"
    get_document.styles["Normal"].font.size = Pt(12)

    # Adding the report heading 

    get_document.add_heading("Weather Report", level=0)

    get_document.add_paragraph(
        f"This is prepared for: {getting_user_name}\n"
        f"The Location is at: {find_city}, {get_country}\n"
        f"The date is download on: "
        f"{datetime.now().strftime("%Y-%m-%d")}\n"
        f"The data source is taken from: Open-Metro Weather API Application"
    )

# Adding the weather table into the word document for the weather report

    current_section = weather_info_section.get("current", {})
    current_units_info = weather_info_section.get("current_units", {})

    get_document.add_heading("Current Conditions", level=1)

    current_table_info = get_document.add_table(rows=1, cols=2)

    current_table_info.style = "Table Grid"

    current_table_info.rows[0].cells[0].text = "Measurement"
    current_table_info.rows[0].cells[1].text = "Value"


# Creating the current weather rows 

    current_rows_info = [
        (
            "Condition",
            weather_description(
                current_section.get("weather_code")
            )
        ),

        (
            "Temperature",
            f"{current_section.get('temperature_2m', 'N/A')}"
            f"{current_units_info.get('temperature_2m', '')}"
        ),

        (
            "Feels Like",
            f"{current_section.get('apparent_temperature', 'N/A')}"
            f"{current_units_info.get('apparent_temperature', '')}"
        ),

        (
            "Humidity",
            f"{current_section.get('relative_humidity_2m', 'N/A')}"
            f"{current_units_info.get('relative_humidity_2m', '')}"
        ),

        (
            "Wind speed",
            f"{current_section.get('wind_speed_10m', 'N/A')}"
            f"{current_units_info.get('wind_speed_10m', '')}"
        )
    ]


    # Adding the following rows to the table 

    for label_section, value_rows in current_rows_info:
        cells_section = current_table_info.add_row().cells
        cells_section[0].text = label_section
        cells_section[1].text = str(value_rows).strip()


    # Adding the daily forecast table 

    daily_weather_section = weather_info_section.get("daily", {})
    daily_units_section = weather_info_section.get("daily_units", {})

    get_document.add_heading("Daily Forecast", level=1)

    forecast_table_section = get_document.add_table(
        rows=1,
        cols=7
    )

    forecast_table_section.style = "Table Grid"


    # Adding the table headings foe daily weather forecast

    headings_daily_section = [
        "Date",
        "Conditions",
        "Maximum",
        "Minimum",
        "Rain Chance",
        "Maximum wind",
        "Sunrise/ Sunset"
    ]

    for index_section, headings in enumerate(headings_daily_section):
        forecast_table_section.rows[0].cells[index_section].text = headings


# Adding the weather forecast values

    dates_section = daily_weather_section.get("time", [])

    for index_section, date_values_section in enumerate(dates_section):
        
        row_section = forecast_table_section.add_row().cells

        row_section[0].text = date_values_section

        row_section[1].text = weather_description(
            daily_weather_section["weather_code"][index_section]
        )

        row_section[2].text = (
            f"{daily_weather_section['temperature_2m_max'][index_section]}"
            f"{daily_units_section.get('temperature_2m_max', '')}"
        )

        row_section[3].text = (
            f"{daily_weather_section['temperature_2m_min'][index_section]}"
            f"{daily_units_section.get('temperature_2m_min', '')}"
        )

        row_section[4].text = (
            f"{daily_weather_section['precipitation_probability_max'][index_section]}"
            f"{daily_units_section.get('precipitation_probability_max', '')}"
        )

        row_section[5].text = (
            f"{daily_weather_section['wind_speed_10m_max'][index_section]}"
            f"{daily_units_section.get('wind_speed_10m_max', '')}"
        )

        sunrise = daily_weather_section["sunrise"][index_section].split("T")[-1]
        sunset = daily_weather_section["sunset"][index_section].split("T")[-1]

        row_section[6].text = f"{sunrise} / {sunset}"

# Saving the Word document for the weather report
# Creating the word document function

    get_document.add_paragraph(
            "The following weather data is provided by Open-Metro."
            "The following forecasts may change due to new information when avaiable"
        )

    get_document.save(report_section_info_path)

    logging.info(
            "Created the weater report: %s",
            os.path.basename(report_section_info_path)
        )

    return report_section_info_path
```

- The following function `report_section_info_path` is gathering following information as follows, the user name, the information of the location that the user has inputted and the weather information as the user requested based on their input.

- Based on the location information that is usally taken towards the API result.

- Country is usally handled when utilising by applying the `.get` due to when retriving the information it can indicate as missing or not found.

- The following path is generated then a new Word Document is created.

- Applied the custom font and font size as well as added the heading, paragraph in which implements the report details based on the weather API data taken from Open Metro.

- Applied the weather measurements for current weather the units.

- Added the section heading along with the table with two coloumns that was produced along with the table styling and headings within the table.

- Within `current_rows_info`, `f"{current_section.get('temperature_2m', 'N/A')}"` explains the following when returning the temperature it does occur ad well as the return `"N/A"`.

- The following loop method creates an item within the Word Document table.

- The daily weather information is gathered as well as the heading, seven column table is created within the word document.

- The loop is added for the headings whereas the `enumerate()` gives the position and the value.

- Retriving the `dates_section` the loop runs once when the dates are gathered, a new table is created as well as date, followed by the weather condition is being converted as well as the maximum temperature is added towards the word document report.

- The API returns the sunset and sunrise in which displays the time based on the selected city.

## Moving the following word document weather report to logs folder

```python
# Creating a function in order to move word document for weather report to logs folder

def moving_weather_report_to_logs_folder(report_section_info_path):
    
    final_destination = os.path.join(
    LOGS_DIR,
    os.path.basename(report_section_info_path)
    )

    shutil.move(
        str(report_section_info_path),
        str(final_destination)
    )

    logging.info(
        "Successfully moved report into the logs folder: %s",
        os.path.basename(final_destination)
    )

    return final_destination
```

When compeleting the word document report the following report will be moved to logs folder that the filename that was created whenever a log entry is created.

## Running the complete program for the weather forecast application

```python
# Printing the comlpete weather api program as follows:

try: 
    print("Weather API Forecast Report Generator")
    print("=====================================")

    getting_user_name = getting_empty_input(
        "Please enter your name: "
    )

    getting_location = getting_empty_input(
        "Please enter a city: "
    )


    print("=====================================")
    
    forecast_days_section = getting_weather_forecast_days()

    print("\nFinding the location as follows....")


    location_section = geocode_city_coordinates(getting_location)

    print("Downloading the weather data....")

    print("=====================================")


    weather_info_section = fetching_weather_info(
        latitude=location_section["latitude"],
        longitude=location_section["longitude"],
        forecast_days=forecast_days_section
    )


    # Displaying the current weather

    current_section = weather_info_section.get("current", {})
    current_units_info = weather_info_section.get("current_units", {})

    print("\nThe Current Weather is:")
    print("=====================================")

    print(f"Location: {location_section['name']}, "
          f"{location_section.get('country')}"
         )

    print(f"Conditions: "
          f"{weather_description(current_section.get('weather_code'))}"
         )

    print(f"Temperature: "
          f"{current_section.get('temperature_2m', 'N/A')}"
          f"{current_units_info.get('temperature_2m', '')}"
         )

    print(f"Humidity: "
          f"{current_section.get('relative_humidity_2m', 'N/A')}"
          f"{current_units_info.get('relative_humidity_2m', '')}"
         )

    print(f"Wind Speed: "
          f"{current_section.get('wind_speed_10m', 'N/A')}"
          f"{current_units_info.get('wind_speed_10m', '')}"
         )


# Creating and moving thw word document weather report

    print("\nCreating the Word Document Weather Report...")

    report_path_section_info_path = creating_word_doc_report(
        getting_user_name,
        location_section,
        weather_info_section
        )

    final_section = moving_weather_report_to_logs_folder(
        report_path_section_info_path
    )

    print("===========================================================================")

    print("\nCongrats!")

    print("The following weather data information has been completed successfully.")

    print("===========================================================================")

    print(
        f"The completed Word Document Report has been saved in the following section:"
        f"{final_section}"
    )

    # Handling the API Errors 

except requests.RequestException as error_api_statement:

    logging.exception(
        "The Weather API request has failed. Please try again."
    )

    print(f"The Weather API error is as follows: {error_api_statement}")

# Handling any invaild city errors

except ValueError as city_error:

    logging.exception("Invalid location or value. Please Try again.")
    print(f"Input Error: {city_error}")


# Handling any unexpected errors

except Exception as unexpected_error:

    logging.exception("There is an unexpected program in your error. Please review your code.")
    print(f"Unexpected error that has been occurred: {unexpected_error}")


# Viewing the completed weather api reports 

finished_reports = [
    os.path.join(LOGS_DIR, file_report)
    for file_report in os.listdir(LOGS_DIR)
    if file_report.endswith(".docx")
]

if finished_reports:
    print("Sucessfully completed weather api reports:")

    for finished_report in finished_reports:
        
        print("-", os.path.basename(finished_report))

else: 
        print("The following compelted reports were not found.")
```
